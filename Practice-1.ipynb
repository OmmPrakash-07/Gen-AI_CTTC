```python
# Import Libraries
import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

from sklearn.metrics import mean_squared_error, r2_score
```

```python
# Load Dataset
df = pd.read_excel("/content/ENB2012_data.xlsx")
print(df.head())
```

```python
# Check Dataset
print(df.shape)
```

```python
print(df.info())
```

```python
print(df.describe())
```

```python
print(df.isnull().sum())
```

```python
df = df.iloc[:, :10]
```

```python
# Select Features and Target
# Input Features
X = df[['X1','X2','X3','X4','X5','X6','X7','X8']]
```

```python
y = df['Y1']
```

```python
y = df['Y2']
```

```python
# Split Dataset
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42
)
```

```python
# Feature Scaling
scaler = StandardScaler()

X_train = scaler.fit_transform(X_train)

X_test = scaler.transform(X_test)
```

```python
# Create ANN Model
model = Sequential()
```

```python
# First Hidden Layer
model.add(Dense(
    units=64,
    activation='relu',
    input_shape=(8,)
))
```

```python
# Second Hidden Layer
model.add(Dense(
    units=32,
    activation='relu'
))
```

```python
# Third Hidden Layer
model.add(Dense(
    units=16,
    activation='relu'
))
```

```python
# Output Layer
model.add(Dense(
    units=1
))
```

```python
# Compile Model
model.compile(
    optimizer='adam',
    loss='mse',
    metrics=['mae']
)
```

```python
# Train Model
history = model.fit(
    X_train,
    y_train,
    epochs=100,
    batch_size=16,
    validation_split=0.2
)
```

```python
# Predict
prediction = model.predict(X_test)
```

```python
# Evaluate Model
rmse = np.sqrt(mean_squared_error(y_test, prediction))

r2 = r2_score(y_test, prediction)

print("RMSE :", rmse)

print("R2 Score :", r2)
```

```python
# Model Summary
model.summary()
```

```python
# Save Model
model.save("Energy_Efficiency_ANN.h5")
```
