```python
#encoding
from sklearn.preprocessing import LabelEncoder
a = LabelEncoder()

#apply -> fruit
frut.fruit=a.fit_transform(frut['fruit'])
```

```python
# Input/output creation
ip = frut.drop('fruit', axis=1)
op = frut['fruit']
```

```python
ip
```

```python
#to categorical
from keras.utils import to_categorical
op = to_categorical(op,4)
```

```python
op
```

```python
from sklearn.model_selection import train_test_split

x_train, x_test, y_train, y_test = train_test_split(
    ip,
    op,
    test_size=0.30,
    random_state=42,
    stratify=frut["fruit"]
)
```

```python
from re import X
print(x_train.shape)
print(x_test.shape)
print(y_train.shape)
print(y_test.shape)
```

```python
#scaler methods
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()
x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
x_train
```

```python
#model create
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

#model
model = Sequential()

#input layer
model.add(Dense(30, input_dim=4, activation='relu'))

#hidden layer
model.add(Dense(128, activation='relu'))
model.add(Dense(64, activation='relu'))
model.add(Dense(32, activation='relu'))

#output layer
model.add(Dense(4, activation='softmax'))

# Compile
model.compile(
    optimizer=Adam(learning_rate=0.01),
    loss="categorical_crossentropy",
    metrics=["accuracy"]
)
```

```python
model.fit(x_train,y_train,validation_split=0.10,epochs=20)
```

```python
#pred
pred = model.predict(x_test)
pred # model predicted output values
```

```python
y_test
```

```python
import numpy as np
b = np.argmax(pred, axis=1)
b
```

```python
c = np.argmax(y_test, axis=1)
c
```

```python
from sklearn.metrics import mean_squared_error, accuracy_score

ac = accuracy_score(b,c)
mse = mean_squared_error(b,c)
ac, mse
```

```python
import numpy as np

# Predict probabilities
pred = model.predict(x_test)

# Convert probabilities to class numbers
predicted_classes = np.argmax(pred, axis=1)

# Convert class numbers back to fruit names
predicted_fruits = a.inverse_transform(predicted_classes)

print(predicted_fruits)
```

```python
actual_classes = np.argmax(y_test, axis=1)
actual_fruits = a.inverse_transform(actual_classes)

for actual, predicted in zip(actual_fruits, predicted_fruits):
    print("Actual:", actual, "| Predicted:", predicted)
```
